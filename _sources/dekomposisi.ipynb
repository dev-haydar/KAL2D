# DEKOMPOSISI QR

Algoritma QR adalah metode numerik yang paling umum digunakan untuk menemukan semua 
nilai eigen dari sebuah matriks. Algoritma ini sangat stabil dan efisien untuk matriks berukuran 
sedang hingga besar.


## Contoh Soal Dekomposisi QR

Disini terdapat matriks 4x4 sebagai contoh soal
$$A = \begin{bmatrix} 5 & 12 & 3 & 8 \\ 0 & 15 & 7 & 2 \\ 11 & 4 & 19 & 1 \\ 6 & 9 & 14 & 10 \end{bmatrix}$$
