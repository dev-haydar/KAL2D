# Rumus Umum

Determinan dengan ekspansi baris:

$$
\det(A)=\sum (-1)^{i+k} a_{ik} M_{ik}
$$

dengan:

- $a_{ik}$ = elemen matriks
- $M_{ik}$ = minor



# 1. Matriks 2×2

$$
A=
\begin{bmatrix}
-7 & -5\\
1 & 4
\end{bmatrix}
$$

Ekspansi baris pertama:

$$
\det(A)=(-1)^{1+1}(-7)M_{11}+(-1)^{1+2}(-5)M_{12}
$$

Minor:

$$
M_{11}=4
$$

$$
M_{12}=1
$$

Perhitungan:

$$
=(+1)(-7)(4)+(-1)(-5)(1)
$$

$$
=-28+5
$$

$$
=-23
$$

**Hasil:**

$$
\det(A)=-23
$$



# 2. Matriks 3×3

$$
A=
\begin{bmatrix}
0&2&-3\\
1&-2&-1\\
0&0&1
\end{bmatrix}
$$

Ekspansi baris pertama:

$$
\det(A)
=(+1)(0)M_{11}
+(-1)(2)M_{12}
+(+1)(-3)M_{13}
$$

Minor:

$$
M_{11}
=
\begin{vmatrix}
-2&-1\\
0&1
\end{vmatrix}
=-2
$$

$$
M_{12}
=
\begin{vmatrix}
1&-1\\
0&1
\end{vmatrix}
=1
$$

$$
M_{13}
=
\begin{vmatrix}
1&-2\\
0&0
\end{vmatrix}
=0
$$

Sehingga:

$$
\det(A)=0-2+0=-2
$$

**Hasil:**

$$
\det(A)=-2
$$



# 3. Matriks 4×4

$$
A=
\begin{bmatrix}
1&-3&1&1\\
-3&1&1&1\\
1&1&-3&1\\
1&1&1&-3
\end{bmatrix}
$$

Ekspansi baris pertama:

$$
\det(A)
=
(+1)(1)M_{11}
+(-1)(-3)M_{12}
+(+1)(1)M_{13}
+(-1)(1)M_{14}
$$

Dari perhitungan minor diperoleh:

$$
M_{11}=16
$$

$$
M_{12}=-16
$$

$$
M_{13}=16
$$

$$
M_{14}=-16
$$

Substitusi:

$$
\det(A)
=
(1)(16)
+(3)(-16)
+(1)(16)
+(-1)(-16)
$$

$$
=16-48+16+16
$$

$$
=0
$$

**Hasil:**

$$
\det(A)=0
$$



# Invers Matriks Dengan Adjoin

Rumus:

$$
adj(A)_{ij}=(-1)^{i+j}M_{ji}
$$

$$
A^{-1}
=
\frac{1}{det(A)}
adj(A)
$$


```

import numpy as np

A = np.array([
[-7,-5],
[1,4]
])

print("Determinan:")
print(np.linalg.det(A))

print("\nInverse:")
print(np.linalg.inv(A))
```
